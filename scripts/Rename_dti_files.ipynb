```python
import os

# Répertoire contenant les fichiers
dossier = "/media/nicolascatz/ssdcalcul/Projets/CRPN_DTI_AGING_DPAKNC/data/Corrected_Tractometry2025"

# Parcourir tous les fichiers du dossier
for nom_fichier in os.listdir(dossier):
    # Vérifier si le nom du fichier contient "MergedTOMs"
    if "MergedTOMs" in nom_fichier:
        # Construire le nouveau nom de fichier
        nouveau_nom = nom_fichier.replace("MergedTOMs", "mergedTOMs")
        # Renommer le fichier
        os.rename(
            os.path.join(dossier, nom_fichier),
            os.path.join(dossier, nouveau_nom)
        )
        print(f"Renommé : {nom_fichier} -> {nouveau_nom}")
```

```python

# Parcourir tous les fichiers du dossier
for nom_fichier in os.listdir(dossier):
    # Vérifier si le nom du fichier se termine par "_last.csv"
    if nom_fichier.endswith("_last.csv"):
        # Construire le nouveau nom de fichier (enlever "_last")
        nouveau_nom = nom_fichier.replace("_last.csv", ".csv")
        # Chemin complet des fichiers
        ancien_chemin = os.path.join(dossier, nom_fichier)
        nouveau_chemin = os.path.join(dossier, nouveau_nom)
        # Renommer le fichier
        os.rename(ancien_chemin, nouveau_chemin)
        print(f"Renommé : {nom_fichier} -> {nouveau_nom}")
```
